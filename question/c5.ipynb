```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/youngwoos/Doit_Python/refs/heads/main/Data/midwest.csv')
df.head()
```

```python
# 1. 데이터 파악
display(df.tail())
print('shape:');display(df.shape)
print('info');display(df.info())
print('describe:');display(df.describe())
```

- 차원: 437x28
- null 없음

```python
# 2. poptotal->total, popasian -> asian
df = df.rename(columns={'poptotal':'total',
                        'popasian':'asian'})
display(df.head())
display(df[['total', 'asian']].head())
```

```python
# 전체 대비 asian 인구 백분율
df['ratio_of_asian'] = df['asian'] / df['total'] * 100
display(df.head())
display(df[['ratio_of_asian']].head())

# hist
df['ratio_of_asian'].plot.hist()
```

```python
# 아시아 인구 백분율 평균 초과시 large, 아니면 small
import numpy as np
asian_avg= sum(df['ratio_of_asian'])/len(df)
print(asian_avg)
df['avg_of_asian'] = np.where(
    df['ratio_of_asian'] > asian_avg,
    'large', 'small'
)
df[['asian', 'ratio_of_asian', 'avg_of_asian']].sample(10)
```

```python
# 아시안 빈도표와 막대그래프
freq_of_asian = df['avg_of_asian'].value_counts()
display(freq_of_asian)
freq_of_asian.plot.bar()
```
