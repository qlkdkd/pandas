```python
import pandas as pd
exam = pd.read_csv('https://raw.githubusercontent.com/youngwoos/Doit_Python/refs/heads/main/Data/exam.csv')
exam.head()
```

```python
'''같음, 같지 않음'''
# 1반 학생들만 추출
print('nclass=1');display(exam.query('nclass == 1'))

# 2반 학생
print('nclass==2');display(exam.query('nclass==2'))

# 1반 학생 아님
print('nclass!=1');display(exam.query('nclass!=1').head())


'''초과, 미만, 이상, 이하'''
print('math>50');display(exam.query('math>50').head())
print('math<50');display(exam.query('math<50').head())
```

```python
# 조건 여러개
print('nclass=1 and math>=50')
display(exam.query('nclass==1 & math >= 50'))

print('math>=90 or english>=90')
display(exam.query('math>=90 | english>=90'))

# 목록에 해당하는 행
print('nclass: 1, 3, 5')
exam.query('nclass in [1, 3, 5]')
```

```python
# 추출한 행으로 데이터 만들기
nclass1 = exam.query('nclass==1')
nclass2 = exam.query('nclass==2')

print(nclass1['math'].mean(), '\n', nclass2['math'].mean())
```

```python
# 문자 변수를 이용해 조건에 맞는 행 추출
df = pd.DataFrame({
    'sex': ['F', 'M', 'F', 'M'],
    'country': ['Korea', 'China', 'Japan', 'USA']
})
display(df)

# 조건 추출
print('sex==F and country==Korea')
display(df.query('sex=="F" & country == "Korea"'))

# 외부 변수를 이용해 추출
var = 3
display(exam.query('nclass==@var'))
```
